# 05. Evaluasi Kritis: Baseline Mayoritas vs Top-3 Hit Rate
Analisis integritas akademik terhadap klaim performa model.


```python
import pandas as pd
import numpy as np

df = pd.read_csv('ml/data/processed/dataset_bersih.csv')
total = len(df)

# Baseline 1: Mayoritas
majority_class = df['product_category'].value_counts().index[0]
majority_share = (df['product_category'] == majority_class).mean()
print(f'Kelas Mayoritas: {majority_class} ({majority_share*100:.2f}%)')
print(f'Akurasi KNN    : 34.89% (Berada di bawah baseline!)')

# Baseline Top-3 Populer
top3 = df['product_category'].value_counts().nlargest(3).index.tolist()
top3_share = df['product_category'].isin(top3).mean()
print(f'\nTop-3 Menu Terpopuler: {top3}')
print(f'Baseline Top-3 Sederhana : {top3_share*100:.2f}%')
print(f'Top-3 Hit Rate KNN       : 78.83% (Berada di bawah baseline top-3!)')

print('\n--- KESIMPULAN JUJUR AKADEMIK ---')
print('Model KNN tidak melampaui baseline frekuensi sederhana pada tugas klasifikasi kategori.')
print('Klaim 78.83% TIDAK BOLEH disajikan sebagai keunggulan model.')
```
