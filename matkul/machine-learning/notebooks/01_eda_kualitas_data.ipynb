Praktikum Machine Learning — Cabang Lower Manhattan (store_id=5)


```python
import os
import pandas as pd
import numpy as np

# Path dataset relatif dari root repositori
RAW_PATH = os.path.join('dataset', 'Coffee Shop Sales.csv')
CLEAN_PATH = os.path.join('ml', 'data', 'processed', 'dataset_bersih.csv')

print('Memeriksa keberadaan dataset mentah...')
assert os.path.exists(RAW_PATH), 'Dataset mentah tidak ditemukan!'
raw_df = pd.read_csv(RAW_PATH)
print(f'Total baris mentah (3 cabang): {len(raw_df):,}')

print('\nMemeriksa dataset bersih Lower Manhattan...')
clean_df = pd.read_csv(CLEAN_PATH)
print(f'Total baris Lower Manhattan: {len(clean_df):,}')
print('Kategori produk:', clean_df['product_category'].nunique())
```

```python
# Distribusi Kategori Produk
dist = clean_df['product_category'].value_counts()
prop = clean_df['product_category'].value_counts(normalize=True) * 100
df_dist = pd.DataFrame({'Jumlah': dist, 'Persentase (%)': prop.round(2)})
print(df_dist)
```
